# Phase 9: Smith-Waterman CPU Implementation

This notebook validates and benchmarks the CPU Smith-Waterman local alignment baseline before future CUDA acceleration.

Smith-Waterman finds the best local alignment region between two sequences. This differs from Needleman-Wunsch global alignment, which aligns full sequences end-to-end.
